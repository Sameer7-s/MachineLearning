```python
import numpy as np 
import pandas as pd 
df = pd.read_csv("employees_100.csv")
df
```

```python
df.head()
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
df['city'].unique()
```

```python
df['department'].nunique()
```

```python
df.groupby("city")['salary'].mean()
```

```python
df.groupby("city")
```

```python
df.groupby("city")['salary'].max()
df.groupby("city")['age'].min()
df.groupby("city")['age'].median()
```

```python
df.groupby(['department','city'])['salary'].mean()
```

```python
## multiple col at once 
df.groupby("department")[["salary" , "performance_score" ]].mean()
```

```python
## multiple aggregation with agg()
df.groupby("department").agg({
    "salary" : ["mean" , "max"] ,
    "performance_score" : "mean" , 
    "employee_id" : "count"
    
})
```

```python
department_summary = df.groupby("department").agg(
    average_salary = ("salary" , "mean") ,
    highest_salary = ("salary" , "max") , 
    employee_count = ("employee_id","count")
     
)
department_summary
```
