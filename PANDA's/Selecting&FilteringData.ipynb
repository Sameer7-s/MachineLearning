```python
# location and index location iLOC[]
df = df.set_index("employee_id")
df.head()
```

```python
df.loc[[184]]
```

```python
df.loc[[184 , 154]]
```

```python
## slice out of the row 
df.loc[184 : 154]
```

```python
##iloc 
## works in the indexing 
df.iloc[1:5]
```

```python
df.loc[154:145][["name","city","age" , "salary"]]
```

```python
##with the help of the iloc function 
df.iloc[1:5][["name","city","age" , "salary"]]
```
