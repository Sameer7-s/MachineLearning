```python
import numpy as np 
import pandas as pd 

df = pd.read_csv("employees_100.csv")
df.head()
df.duplicated(subset=["employee_id"]).sum()
df.drop_duplicates(subset=["employee_id"],keep="first",inplace=True)
```

```python
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')

df['annual_salary'] = df['salary'] * 12

df.head()
```

```python
def categorize(num):
    if(num>5):
        return "Experienced candidate"
    elif num<= 5:
        return "learner"

##using the lambda function 

## when we have single condition so we can use this 

# df["salary_level"] = df["salary"].apply(
#     lambda salary: "High" if salary > 70000 else "Normal"
# )
```
