```python
df2 = df.dropna()
df2
```

```python
 ## what if we want to drop the value of city only 
df.dropna(subset=['city'])
```

```python
df.dropna(subset=['city' , 'age'])
```

```python
df.isnull().sum()
```

```python
df['age'].fillna(df['age'] .mean(),inplace = True)
```

```python
df.isnull().sum()
##now age is filled with mean 
```

```python
# df['salary'] = df['salary'].str.replace(" , " , "",regex= False)
# df['salary'] = df['salary'].astype(int) it does not convert the null values to an integer 
# df['salary'].fillna(df['salary'].median() , inplace = True)
```

```python
df['salary'] = df['salary'].str.replace(",", "" , regex=False)
df['salary'] = pd.to_numeric(df['salary'] , errors='coerce')##coerce means skip krna 
df['salary'] = df['salary'].fillna(df['salary'].median())
```

```python
df['salary'].dtype
```

```python
df.isnull().sum()
```

for department (string)

```python
df['department'].value_counts()
```

```python
df['department'] = df['department'].fillna(df['department'].mode()[0])
df['city'] = df['city'].fillna(df['city'].mode()[0])
```

```python
df.isnull().sum()
```
