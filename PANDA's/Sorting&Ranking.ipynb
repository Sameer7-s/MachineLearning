```python
import numpy as np 
import pandas as pd
df = pd.read_csv("employees_100.csv")
df.head()
df.duplicated(subset=["employee_id"]).sum()
df.drop_duplicates(subset=["employee_id"],keep="first",inplace=True)
## Top and bottom records
```

 WHY DO WE NEED THIS

After analyzing data, you usually want the extremes — the highest, lowest, best, or worst performing records, not the whole table in whatever order it happened to load in.
