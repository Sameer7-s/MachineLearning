```python
import pandas as pd 
import matplotlib.pyplot as plt 
df = pd.read_csv("../../datasets/data_visualization_student_dataset.csv")
df
```

```python
department_count = df['Department'].value_counts()
department_count
```

```python
plt.bar(department_count.index , department_count.values)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Number of Students by Department")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    department_count.index,
    department_count.values
)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Number of Students by Department")
plt.xticks(rotation=45)
plt.grid(axis="y" , linestyle = "--" , alpha = 1)

plt.show()
```

```python
plt.hist(df['Marks'])

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

```python
plt.hist(df['Marks'] , bins = 45)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

```python
plt.hist(
    df['Marks'] , 
    bins = 20 , 
    edgecolor = "black"    
    )

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

# Scatter plot

```python
plt.scatter(df.index , df['Marks'])
plt.xlabel("Student Index")
plt.ylabel("Marks")
plt.title("Student marks destribution")
plt.show()
```

```python
plt.scatter(df["Study_Hours"] , df['Marks'])
plt.xlabel("Student Index")
plt.ylabel("Marks")
plt.title("Student marks destribution")
plt.show()
```

```python
department = df["Department"].unique()

colors = {
    "MCA" : "blue",
    "MBA" : "red",
    "BCA" : "yellow",
    "BOT" : "orange",
}

for department in department:
    data = df[df["Department"] == department]
    plt.scatter(
     data["Study_Hours"] , 
    data["Marks"] , 
    color = colors[department] , 
    label = department , 
    alpha = 0.7
        
)

    plt.xlabel("Study Hours")
    plt.ylabel("Marks")

    plt.title("Study Hours vs Marks by Department")

    plt.legend(title = "Department")

    plt.show()
```
