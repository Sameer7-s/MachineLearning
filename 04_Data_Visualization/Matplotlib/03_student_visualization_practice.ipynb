```python
department = df["Department"].unique()

colors = {
    "MCA" : "blue",
    "MBA" : "red",
    "BCA" : "yellow",
    "BOT" : "orange",
}

for department in department:
    data = df[df["Department"] == department]
    plt.scatter(
     data["Study_Hours"] , 
    data["Marks"] , 
    color = colors[department] , 
    label = department , 
    alpha = 0.7
        
)

    plt.xlabel("Study Hours")
    plt.ylabel("Marks")

    plt.title("Study Hours vs Marks by Department")

    plt.legend(title = "Department")

    plt.show()
```

```python

import numpy as np

# Filter students department-wise using loc
mca_marks = df.loc[df["Department"] == "MCA", "Marks"]
mba_marks = df.loc[df["Department"] == "MBA", "Marks"]

# MCA calculations
print("MCA Mean:", mca_marks.mean())
print("MCA Standard Deviation:", mca_marks.std())
print("MCA Students:", len(mca_marks))

print()

# MBA calculations
print("MBA Mean:", mba_marks.mean())
print("MBA Standard Deviation:", mba_marks.std())
print("MBA Students:", len(mba_marks))
```

```python
from scipy.stats import ttest_ind
t_statistics , p_value  = ttest_ind(
    mca_marks,
    mba_marks,
    equal_var = False
)

print(t_statistics)
print(p_value)
```

```python
alpha = 0.05

if(p_value<alpha):
    print("Reject Null Hypothesis ")
    print("There is siginificant evidance of a difference")
else:
    print("Fail to reject null hypothesis")
    print("There is not siginificant evidenace of differnnce")
```
