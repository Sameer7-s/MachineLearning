```python
import numpy as np 
a = np.array([10,20,30,40])
print(a[0])
print(a[1])
print(a[2])
print(a[3])
```

```python
## 2D indexing 
## for accesing the values first put row number and coloumn number
##a[row , coloumn]

a = np.array([
    [1,2,4] 
   ,[8,5,6]
])
print(a[0,0])
print(a[0,1])
print(a[0,2])
print(a[1,0])
print(a[1,1])
print(a[1,2])
##print(a[1,3])  this throws an error that index is out of the bounds
```

```python
##slicing
import numpy as np 
a = np.array([10,20,30,40,50])
print(a)
print(a[1:4]) ##slice from indexing 1 to 4  
print(a[:4])
print(a[2:])
print(a[::-1]) ## Reverse the complete array 

```

```python
##Slicing in the 2d array 
import numpy as np 
a = np.array([
    [10,20,30],
    [40,50,60],
])
print(a)
# print(a[0 : 2 , 1 : 3])
print(a[0:3 ,  1:2])
```
