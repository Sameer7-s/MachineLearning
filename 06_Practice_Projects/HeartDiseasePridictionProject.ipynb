# Bar Graph


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

df = pd.read_csv("heart.csv")

sns.countplot(data=df, x="ChestPainType")

plt.title("Number of Patients by Chest Pain Type")
plt.xlabel("Chest Pain Type")
plt.ylabel("Number of Patients")
plt.show()
```

# Line plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("heart.csv")

# Sort data by Age
df = df.sort_values("Age")

# Create line graph
sns.lineplot(data=df, x="Age", y="RestingBP", errorbar=None)

plt.title("Age vs Resting Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Resting Blood Pressure")
plt.grid(True)
plt.show()
```

# histogram

```python

plt.hist(df["Age"], bins=10, edgecolor="black")

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

# Scatter plot

```python

sns.scatterplot(data=df, x="Age", y="MaxHR")

plt.title("Age vs Maximum Heart Rate")
plt.xlabel("Age")
plt.ylabel("Maximum Heart Rate")
plt.show()
```

# Heat map

```python

corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Heart Dataset Correlation Heatmap")
plt.show()
```

```python
ch_mean = df.loc[df['Cholesterol'] != 0 , 'Cholesterol'].mean()
ch_mean
```

```python
df['Cholesterol'] = df['Cholesterol'].replace(0,ch_mean)
df['Cholesterol'] = df['Cholesterol'].round(2)

```

```python
rest_bp = df.loc[df['RestingBP'] != 0 , 'RestingBP'].mean()
rest_bp
# df['RestingBP'] = df['RestingBP'].replace(0,Res_bp)
```

```python
df['RestingBP'] = df['RestingBP'].replace(0 , rest_bp)
df['RestingBP'] = df['RestingBP'].round(2)
```
