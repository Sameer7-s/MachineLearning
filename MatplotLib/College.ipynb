```python
import pandas as pd 
import matplotlib.pyplot as plt 
df = pd.read_csv("Data_Visualization_Student_Dataset.csv")
df
```

```python
department_count = df['Department'].value_counts()
department_count
```

```python
plt.bar(department_count.index , department_count.values)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Number of Students by Department")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    department_count.index,
    department_count.values
)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Number of Students by Department")
plt.xticks(rotation=0)
plt.grid(axis="y")
```
