```python
import pandas as pd 
import matplotlib.pyplot as plt 
df = pd.read_csv("Data_Visualization_Student_Dataset.csv")
df
```

```python
department_count = df['Department'].value_counts()
department_count
```

```python
plt.bar(department_count.index , department_count.values)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Number of Students by Department")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    department_count.index,
    department_count.values
)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Number of Students by Department")
plt.xticks(rotation=45)
plt.grid(axis="y" , linestyle = "--" , alpha = 1)

plt.show()
```

```python
plt.hist(df['Marks'])

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

```python
plt.hist(df['Marks'] , bins = 45)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

```python
plt.hist(
    df['Marks'] , 
    bins = 20 , 
    edgecolor = "black"    
    )

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```
