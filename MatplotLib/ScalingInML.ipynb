```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

df = pd.read_csv("scaling_pandas_student_dataset.csv")
df
```

```python
minimum = df["Study_Hours_Per_Week"].min()
print(minimum)
```

```python
maximum = df["Study_Hours_Per_Week"].max()
print(maximum)
```

```python
df["Study_Hours_Scaled"] = (
    (df["Study_Hours_Per_Week"] - minimum) /
    (maximum - minimum)
)
```

```python
import pandas as pd 
df = pd.read_csv("scaling_pandas_student_dataset.csv")
numeric_columns = df.select_dtypes(include="number").columns
df[numeric_columns] = (
    (df[numeric_columns] - df[numeric_columns].min()) /
    (df[numeric_columns].max() - df[numeric_columns].min())
)
```

```python
import pandas as pd 
df = pd.read_csv("scaling_pandas_student_dataset.csv")
df
```

```python
mean = df["Study_Hours_Per_Week"].mean()
print(mean)
```

```python
stand_dev = df["Study_Hours_Per_Week"].std()
print(stand_dev)
```

```python
df["Study_Hours_Z"] = (
    (df['Study_Hours_Per_Week'] - mean) / stand_dev
)
```
