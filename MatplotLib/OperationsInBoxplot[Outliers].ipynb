```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("student_information_with_outliers - student_information_with_outliers.csv")
df.boxplot(column = "CGPA")

plt.title("CGPA BoxPlot")
plt.show()
```

```python
df.boxplot(
    column=[
        "Age",
        "Attendance_Percentage",
        "Internal_Marks",
        "CGPA"
    ]
)
plt.show()
```

```python
Q1 = df["CGPA"].quantile(0.25)
Q3 = df["CGPA"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 =" , Q1)

print("Q3 =" , Q3)
print("IQR = " , IQR)
```

Calculating the lower and the upper bond in box plot 


```python
lower_bound = Q1 - 1.5 *IQR
upper_bound = Q3 + 1.5 *IQR
print("Lower_bound" , lower_bound)
print("Upper_bound" , upper_bound)
```

```python
outlier = df[
    (df["CGPA"]<lower_bound) |
    (df["CGPA"]>upper_bound)
]
print(outlier)
```
