```python
import numpy as np
a = np.array([1,2,3])
b = np.array([4,5,6])

print(a + b)
print(a * b)
print(a ** b)
```

STATISTICAL AGGREGATION 

```python
import numpy as np 
data1 = np.array([[1,2] ,[3,4]])
data2 = np.array([1,2,3])
print(np.sum(data1))
print(np.sum(data2))
print(np.mean(data1))
print(np.max(data1))
print(np.min(data1))
```

```python
##use of axis 
data1 = np.array([[1,2] 
                  ,[3,4]])
data2 = np.array([1,2,3])
print(np.sum(data1 , axis=0))
print(np.sum(data1 , axis=1))
##axix = 0   column 
##axis = 1 row 
```

```python
## RESHAPING 
## Covert the any dim array to any other dimenstionality 
flat = np.array([1,2,3,4,5,6])
matrix = flat.reshape(3,2)
print(matrix)
```

```python
##Boolean Masking 
arr = np.array([12,34,5,223,1])
mask = arr>20
print(mask)
```
